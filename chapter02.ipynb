# 동적 타이핑과 재할당
파이썬은 타입을 선언하지 않습니다. 실행 할 때 **가리키는 객체**가 타입을 정해요.
재할당은 이름표를 다른 객체로 옮기는 것이라 type()과 id()가 함께 변합니다. 대신 타입 실수는 실행해야 드러납니다.

````python
x = 10
print(type(x))
print(id(x))

x = "Hello"
print(type(x))
print(id(x))

"10" + 5
````

# 기본 자료형 네 가지

### int
* 정수 25 -5 0

### float
* 실수 175.5 3.14

### str
* 문자열 "Sparta" "25"

### bool
* 불리언 True False
* bool(값) 채워져 있으면 True
    -1 0.1 "Hi" [0]
* bool(값) 비워져 있으면 False
    0 0.0 "" [] None
* bool("0"), bool(" "), bool("False")
    True, True, True
    문자열은 글자의 뜻이 아니라 비었는지만 봄

```python
type(10), type(10,0) # int, float
10 == 10.0 # True(값은 같음)
type("25") # <class 'str'> 따옴표 = 문자열
0.1 + 0.2 # 0.3000000000000000004
```
