## 함수 (Function)

- 반복되는 로직을 합쳐놓은 문법
- 재사용 높일 수 있는 코드 기법
- 단독사용 가능 print(), len(), ...
- 함수명 컨벤션 : snake_case
- 매개변수 (Parameter)
```python
def 함수이름 (매개변수, 매개변수 ...):
    실행할 로직 # 1
    실행할 로직 # 2
    실행할 로직 # 3
```


### 함수 정의

````python
def say_hello(name, age, point = 1000:)  # default 값은 뒤에서 선언
    print(f"안녕하세요. 저는 {name} 나이는 {age}입니다.")
    print(f"초기 지급 포인트 {point}점 지급 되었습니다.")

say_hello("이수환", 36, 2000)
````

### 가변 인수

````python
def add_all(*numbers): # 매개변수 (Prameter)
    print(f"입력된 튜플 : {numbers}")
    total = sum(numbers)

    print(f"합계 결과 : {total}")
add_all(1, 2, 3) # 인수 (argument)
add_all(1, 2, 3, 4, 5)
add_all()
````

### 딕셔너리 가변 인수

#### kwargs

````python
def user_info(**data):
    print(f"전달된 딕셔너리 구조: {data}")
    for key, value in data.items():
        print(f"{key} : {value}")

# 함수 호출

user_info(name = "스파르타", level = 5, active = True)
````

#### return - 결괏값 되돌리기

````python
# 하나의 결괏값 반환

def check_ord_number(number):
    if number > 0:
        return "양수"
    else:
        return "음수"
print(check_ord_number(10))
print(check_ord_number(-20))
````

````python
# 여러개 결괏값 반환

def calc_value(a, b):
    plus = a + b
    minus = a - b
    return plus, minus # 튜플로 리턴

result1, result2 = calc_value(10, 5)
print(f"덧셈 : {result1}, 뺄셈 : {result2}")

result = calc_value(10, 5)
print(f"튜플 : {result}")
````

````python
# 평균 계산해서 반환하는 함수

def get_average(*scores):
    if len(scores) == 0:
        return 0

    # if not scores:
    #     return 

    total = sum(scores)
    avg = total / len(scores)
    return avg

result1 = get_average(80, 90, 100)
result2 = get_average(75, 80, 92, 55)
result3 = get_average()

print(f"3과목 평균 값 : {result1}")
print(f"5과목 평균 값 : {result2}")
````

#### 미니 계산기, 사칙 연산기

````python
def add(a, b):
    total = int(a) + int(b)
    return total

def minus(a, b):
    total = int(a) - int(b)
    return total

def multiply(a, b):
    total = int(a) * int(b)
    return total

def divide(a, b):
    if b == 0:
        return "0으로 나눌 수 없습니다."
    total = int(a) / int(b)
    return total

a = input("첫번째 숫자: ")
b = input("두번째 숫자: ")

result1 = add(a, b)
result2 = minus(a, b)
result3 = multiply(a, b)
result4 = divide(a, b)

print(f"덧셈 연산 {a} + {b}: {result1}")
print(f"뺄셈 연산 {a} - {b}: {result2}")
print(f"곱셈 연산 {a} * {b}: {result3}")
print(f"나눗셈 연산 {a} / {b}: {result4}")


print(f"나눗셈 연산 {a} / {b}: {divide(int(a), int(b))}")

````

## 변수 스코프 (유효범위)

- 전역 변수 : 함수 밖에서 선언하는 변수, 어디서나 접근이 가능하다
- 지역 변수 : 함수내에 선언하는 변수, 함수 안에서만 접근이 가능하다

### 지역 변수

````python
def local():
    print(local_var)
    local_var = 1 # 지역 변수

local()
# print(local_var) # 오류 발생
````

###  전역변수

````python
message = "안녕하세요. Kant 몰에 오신 것을 환영합니다."

def greeting():
    message = "장난감 코너는 3층에 있습니다." # 지역 변수
    print(message) # 지역 변수가 우선된다.

greeting()

print(message) # 전역 변수
````

````python
value = "전역 데이터"

def outer_function():
    value = "지역 변수 1"

    def inner_function():
        value = "지역 변수 2"
        print(value)
    inner_function()
    print(value)

    
outer_function()
print(value)
````

````python
# global 키워드

counter = 0

def inc_counter():
    # counter = 0
    global counter # 전역 변수를 참조하겠다
    counter += 1

inc_counter()
inc_counter()
print(counter)
````

## Lambda Expression 람다식
일종의 함수이다  
적절하게 사용한다, 많이 쓰면 가독성이 떨어진다

````python
# lambda 매개변수: 로직

# 일반 함수

def add_five(x):
    return x + 5

print(add_five(10))

# lambda

result = lambda x: x + 5
print(result(20))
````

### map
모든 요소를 반환

````python
numbers = [1, 2, 3, 4, 5]

result = list(map(lambda x: x + 10, numbers))
print(f"계산 결과: {result}")

print([x + 10 for x in numbers])
````

### filter


````python
filter_result = list(filter(lambda x: x % 2 == 0, numbers))
print(f"짝수 결과: {filter_result}")

print([x for x in numbers if x % 2 == 0])
````

### 정렬

````python
players = [("손흥민", 95),
            ("김민재", 60),
            ("이강인", 80)]

print(sorted(players, key = lambda p: p[1]))
print(sorted(players, key = lambda p: p[1], reverse = True))
print(max(players, key = lambda p: p[1]))
````
